```python
from sympy import MatrixSymbol, diff, symbols, trace, Eq



n, m = symbols('n m')
Q = MatrixSymbol('Q', n, n) # Process noise covariance
R = MatrixSymbol('R', m, m) # Measurement noise covariance
P = MatrixSymbol('P', n, n) # Estimate error covariance
P_pred = MatrixSymbol('\overline{P}', n, n) # Estimate error covariance
w = MatrixSymbol('w', n, 1) # Process noise
v = MatrixSymbol('v', m, 1) # Measurement noise
x_hat = MatrixSymbol('\overline{x}', n, 1) # Refined state
x = MatrixSymbol('x', n, 1) # Real state
z = MatrixSymbol('z', m, 1) # Measurement
K = MatrixSymbol('K', n, m) # Kalman gain
F = MatrixSymbol('F', n, n) # State transition model
H = MatrixSymbol('H', m, n) # Observation model
I = MatrixSymbol('I', n, n) # Identity matrix
zero_vec = MatrixSymbol(r"\hat{0}", n, 1) # Zero vector
```

```python
process = F * x + w
measurement = H * x + v
```

```python
Eq(x, process)
```

```python
Eq(x, x_hat + K * (measurement - H * x_hat))
```

```python
Eq(x, x_hat + K * v - K * H * (x - x_hat))
```

```python
Eq(x - x_hat - K * v - K * H * x_hat + K * H * x, zero_vec)
```

```python
gaol_eq = Eq(P, P_pred - K * H * P_pred - P * H.transpose() * K.transpose() + K * (H * P_pred * H.transpose() + R) *  K.transpose())
gaol_eq
```

```python
diff(gaol_eq.rhs, K)
```
