```python
import sympy as sp


x0, y0 = sp.symbols('x0 y0')
x1, y1 = sp.symbols('x1 y1')
x2, y2 = sp.symbols('x2 y2')

alpha = sp.symbols('alpha')
beta = sp.symbols('beta')
gamma = sp.symbols('gamma')
```

```python
p0 = sp.Matrix([x0, y0])
p1 = sp.Matrix([x1, y1])
p2 = sp.Matrix([x2, y2])

c0 = p0 * (1 - alpha) + p1 * alpha
c1 = p1 * (1 - beta) + p2 * beta
c2 = p2 * (1 - gamma) + p0 * gamma

perym = sp.sqrt((c0[0] - c1[0])**2 + (c0[1] - c1[1])**2) + \
        sp.sqrt((c1[0] - c2[0])**2 + (c1[1] - c2[1])**2) + \
        sp.sqrt((c2[0] - c0[0])**2 + (c2[1] - c0[1])**2)
perym.simplify()
```

```python
# minimize perym with respect to alpha, beta, gamma
grad = sp.Matrix([sp.diff(perym, alpha), sp.diff(perym, beta), sp.diff(perym, gamma)])
grad.simplify()
grad_zero = sp.Eq(grad, sp.Matrix([0, 0, 0]))
grad_zero
```

```python
# solve the system
solution = sp.solve(grad_zero, (alpha, beta, gamma))
solution
```
